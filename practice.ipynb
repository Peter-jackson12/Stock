```python
customer_a = {"노트북", "키보드", "마우스"}
customer_b = {"키보드", "모니터", "마우스"}

# 두 고객이 모두 가지고 있는 상품
common_items = customer_a & customer_b
print(f"두 고객 모두 가지고 있는 상품: {common_items}")

# 첫 번째 고객만 가지고 있는 상품
only_customer_a = customer_a - customer_b
print(f"첫 번째 고객만 가지고 있는 상품: {only_customer_a}")

# 두 고객 중 하나라도 가지고 있는 상품
all_items = customer_a | customer_b
print(f"두 고객 중 하나라도 가지고 있는 상품: {all_items}")

list_a = [1, 2, 3, 4, 5]
set_a = set(list_a)
print(f"type: {type(set_a)}")
```

```python
orders = [12000, 55000, 32000, 80000, 45000]
for order in orders:
    if order >= 50000:
        print(f"주문 금액: {order}원 ")

def calculate_discount(price, discount_rate):
    if price >= 50000:
        discount = price * discount_rate
        return discount
    else:
        return 0  
```
